```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import os
import spotipy
from spotipy.oauth2 import SpotifyOAuth
from dotenv import load_dotenv
import requests


load_dotenv()

API_KEY = os.getenv("FANART_API_KEY")

print("API-Key geladen:", API_KEY is not None)
```

```python
df_music = pd.read_pickle("../data/processed/df_music_features.pkl")
```

```python
artists = df_music.dropna(subset=["master_metadata_album_artist_name"])

# Hörzeit pro Künstler (in Minuten)
top_artists = (
    artists.groupby("master_metadata_album_artist_name")["ms_played"]
    .sum()
    .sort_values(ascending=False)
    .reset_index()
)

top_artists["minutes_played"] = top_artists["ms_played"] / 60000

# Top 25
top25 = top_artists.head(25)

print(top25[["master_metadata_album_artist_name", "minutes_played"]])
```
